### Import Python Modules

```python
import requests
import json
import pandas as pd
```

### Set File Paths

```python
volume_path = 'nhl_bronze.default.nhl_data'
```

### Retrieve NHL data snapshot

```python


url = "https://api-web.nhle.com/v1/standings/now"
response = requests.get(url)
teams_data = response.json()
teams_data_json = json.dumps(teams_data)
print(type(teams_data_json))


```

### Read JSON file into Delta Table with PySpark

```python
# Flatten the teams_data dictionary to a pandas DataFrame
flat_df = pd.json_normalize(teams_data['standings'])

# Convert the pandas DataFrame to a Spark DataFrame
spark_df = spark.createDataFrame(flat_df)

# Write the Spark DataFrame to a managed table
spark_df.write.format("delta").mode("overwrite").saveAsTable(volume_path)
```
