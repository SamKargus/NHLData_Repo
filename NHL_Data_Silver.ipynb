# Read from Bronze delta table

```python
from pyspark.sql.functions import col
```

```python
df = spark.table("nhl_bronze.default.nhl_data_20260201_214159")
```

```python
df.display()
df.printSchema()
```

```python
# filtered_df = df.select('teamName')
df_filter = df.select(
    col("teamName")["default"].alias("team_name"),
    col("wins"),
    col("losses")
    )

df_filter.display()
df_filter.printSchema()
```
